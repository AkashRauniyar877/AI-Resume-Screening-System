```python
clean_df = df.copy()
```

```python
clean_df.head()
```

```python
clean_df = clean_df.drop(columns=["Resume_html"])
```

```python
clean_df.head()
```

```python
clean_df.columns
```

```python
clean_df = clean_df.rename(columns={
    "Resume_str": "resume_text"
})
```

```python
clean_df.columns
```

```python
print(clean_df["resume_text"].iloc[0])
```

```python
clean_df["resume_text"].isnull().sum()
```

```python
import re
```

```python
def clean_text(text):
    text = text.lower()
    text = re.sub(r'\s+', ' ', text)
    text = re.sub(r'[^a-zA-Z0-9\s]', ' ', text)
    text = text.strip()
    return text
```

```python
clean_df["clean_resume"] = clean_df["resume_text"].apply(clean_text)
```

```python
print("ORIGINAL:")
print(clean_df["resume_text"].iloc[0])

print("\nCLEANED:")
print(clean_df["clean_resume"].iloc[0])
```

```python
clean_df["text_length"] = clean_df["clean_resume"].str.len()
```

```python
clean_df = clean_df[clean_df["text_length"] > 0]
```

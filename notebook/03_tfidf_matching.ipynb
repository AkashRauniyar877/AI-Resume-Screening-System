```python
import pandas as pd
```

```python
df = pd.read_csv("../data/processed/processed_resume.csv")
```

```python
df.shape
```

```python
df[["Category", "processed_resume"]].head()
```

```python
from sklearn.feature_extraction.text import TfidfVectorizer
```

```python
vectorizer = TfidfVectorizer()
```

```python
tfidf_matrix = vectorizer.fit_transform(df["processed_resume"])
```

```python
tfidf_matrix.shape
```

```python
feature_names = vectorizer.get_feature_names_out()
```

```python
feature_names[:50]
```

```python
[word for word in feature_names if word in ["python", "java", "sql", "react", "mongodb", "fastapi"]]
```

```python
len(feature_names)
```

```python
first_resume_vector = tfidf_matrix[0]
```

```python
first_resume_vector
```

```python
first_resume_vector.toarray()
```

```python
job_description = """
We are looking for a Python developer with experience in
Python, SQL, MongoDB, FastAPI, REST API, React and JavaScript.
The candidate should have experience in backend development,
database management and web application development.
"""
```

```python
job_vector = vectorizer.transform([job_description])
```

```python
from sklearn.metrics.pairwise import cosine_similarity
```

```python
similarity_scores = cosine_similarity(
    job_vector,
    tfidf_matrix
)
```

```python
similarity_scores.shape
```

```python
scores = similarity_scores[0]
```

```python
scores[:10]
```

```python
match_percentage = scores * 100
```

```python
match_percentage[:10]
```

```python
results = df.copy()
```

```python
results["match_score"] = match_percentage
```

```python
results = results.sort_values(
    "match_score",
    ascending=False
)
```

```python
results[
    ["ID", "Category", "match_score"]
].head(20)
```

```python
results[
    ["ID", "Category", "match_score"]
].head(10)
```

```python
results["match_score"].describe()
```

```python
results["rank"] = range(1, len(results) + 1)
```

```python
results[
    ["rank", "ID", "Category", "match_score"]
].head(10)
```

```python
job_description_2 = """
We are looking for a data analyst with experience in
Python, SQL, Excel, statistics, data visualization,
Pandas, NumPy and Power BI.
The candidate should be able to analyze datasets,
create reports and generate business insights.
"""
```

```python
job_vector_2 = vectorizer.transform([job_description_2])
```

```python
similarity_scores_2 = cosine_similarity(
    job_vector_2,
    tfidf_matrix
)
```

```python
scores_2 = similarity_scores_2[0]
```

```python
results_2 = df.copy()
results_2["match_score"] = scores_2 * 100
```

```python
results_2 = results_2.sort_values(
    "match_score",
    ascending=False
)
```

```python
results_2[
    ["ID", "Category", "match_score"]
].head(10)
```

```python
results[
    ["ID", "Category", "match_score", "rank"]
].to_csv(
    "../data/processed/tfidf_matching_results.csv",
    index=False
)
```

```python
skills = [
    "python",
    "java",
    "c++",
    "javascript",
    "typescript",
    "html",
    "css",
    "react",
    "node.js",
    "nodejs",
    "express",
    "fastapi",
    "flask",
    "django",
    "sql",
    "mysql",
    "mongodb",
    "postgresql",
    "oracle",
    "redis",
    "git",
    "github",
    "docker",
    "aws",
    "azure",
    "linux",
    "excel",
    "power bi",
    "tableau",
    "pandas",
    "numpy",
    "scikit-learn",
    "tensorflow",
    "pytorch",
    "machine learning",
    "deep learning",
    "natural language processing",
    "nlp",
    "data analysis",
    "data visualization",
    "rest api",
    "rest",
    "api"
]
```

```python
len(skills)
```

```python
def extract_skills(text):
    found_skills = []

    text = text.lower()

    for skill in skills:
        if skill in text:
            found_skills.append(skill)

    # Sort longest skills first
    found_skills.sort(key=len, reverse=True)

    final_skills = []

    for skill in found_skills:
        if not any(skill in existing for existing in final_skills):
            final_skills.append(skill)

    return final_skills
```

```python
text = df["clean_resume"].iloc[0]

extract_skills(text)
```

```python
df["extracted_skills"] = df["clean_resume"].apply(extract_skills)
```

```python
import pandas as pd

tfidf_results = pd.read_csv(
    "../data/processed/tfidf_matching_results.csv"
)

skill_results = pd.read_csv(
    "../data/processed/skill_matching_results.csv"
)

print("TF-IDF:", tfidf_results.shape)
print("Skills:", skill_results.shape)
```

```python
print(tfidf_results["ID"].head())
print(skill_results["ID"].head())
```

```python
print("TF-IDF duplicate IDs:", tfidf_results["ID"].duplicated().sum())
print("Skill duplicate IDs:", skill_results["ID"].duplicated().sum())
```

```python
combined_results = tfidf_results.merge(
    skill_results[
        [
            "ID",
            "matched_skills",
            "missing_skills",
            "skill_coverage"
        ]
    ],
    on="ID",
    how="inner"
)
```

```python
combined_results["overall_score"] = (
    0.5 * combined_results["match_score"]
    + 0.5 * combined_results["skill_coverage"]
)
```

```python
combined_results = combined_results.sort_values(
    "overall_score",
    ascending=False
).reset_index(drop=True)
```

```python
combined_results["overall_rank"] = range(
    1,
    len(combined_results) + 1
)
```

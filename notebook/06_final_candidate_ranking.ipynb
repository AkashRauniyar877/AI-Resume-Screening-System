```python
import pandas as pd
```

```python
tfidf_results = pd.read_csv(
    "../data/processed/tfidf_matching_results.csv"
)

skill_results = pd.read_csv(
    "../data/processed/skill_matching_results.csv"
)

experience_results = pd.read_csv(
    "../data/processed/experience_results.csv"
)

print("TF-IDF:", tfidf_results.shape)
print("Skills:", skill_results.shape)
print("Experience:", experience_results.shape)
```

```python
print("TF-IDF duplicate IDs:",
      tfidf_results["ID"].duplicated().sum())

print("Skills duplicate IDs:",
      skill_results["ID"].duplicated().sum())

print("Experience duplicate IDs:",
      experience_results["ID"].duplicated().sum())
```

```python
combined_results = tfidf_results.merge(
    skill_results[
        [
            "ID",
            "matched_skills",
            "missing_skills",
            "skill_coverage"
        ]
    ],
    on="ID",
    how="inner"
)

print("After TF-IDF + Skills:", combined_results.shape)
```

```python
combined_results = combined_results.merge(
    experience_results[
        [
            "ID",
            "calculated_experience",
            "experience_score"
        ]
    ],
    on="ID",
    how="inner"
)

print("Final merged shape:", combined_results.shape)
```

```python
print("Number of candidates:", len(combined_results))

print(
    "Duplicate IDs after merge:",
    combined_results["ID"].duplicated().sum()
)

print(
    "Missing experience scores:",
    combined_results["experience_score"].isna().sum()
)
```

```python
combined_results["overall_score"] = (
    0.30 * combined_results["match_score"]
    + 0.40 * combined_results["skill_coverage"]
    + 0.30 * combined_results["experience_score"]
)
```

```python
combined_results = combined_results.sort_values(
    "overall_score",
    ascending=False
).reset_index(drop=True)
```

```python
combined_results["overall_rank"] = range(
    1,
    len(combined_results) + 1
)
```

```python
top_candidates = combined_results[
    [
        "overall_rank",
        "ID",
        "Category",
        "match_score",
        "skill_coverage",
        "experience_score",
        "overall_score",
        "matched_skills",
        "missing_skills"
    ]
].head(20)

top_candidates
```

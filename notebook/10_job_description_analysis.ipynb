```python
import pandas as pd
import re
```

```python
job_description = """
We are looking for a Python Developer with 3+ years of experience
in backend development.

The candidate should have strong knowledge of Python, FastAPI,
REST API, MongoDB, SQL, React and JavaScript.

Experience with Git, Docker and AWS is preferred.

The candidate should have a Bachelor's degree in Computer Science,
Information Technology or a related field.
"""
```

```python
print(job_description)
```

```python
skills = [
    "python",
    "java",
    "c++",
    "javascript",
    "typescript",
    "html",
    "css",
    "react",
    "node.js",
    "nodejs",
    "express",
    "fastapi",
    "flask",
    "django",
    "sql",
    "mysql",
    "mongodb",
    "postgresql",
    "oracle",
    "redis",
    "git",
    "github",
    "docker",
    "aws",
    "azure",
    "linux",
    "excel",
    "power bi",
    "tableau",
    "pandas",
    "numpy",
    "scikit-learn",
    "tensorflow",
    "pytorch",
    "machine learning",
    "deep learning",
    "natural language processing",
    "nlp",
    "data analysis",
    "data visualization",
    "rest api",
    "rest",
    "api"
]
```

```python
def extract_required_skills(text):
    found_skills = []
    text = text.lower()

    for skill in skills:
        if skill in text:
            found_skills.append(skill)

    found_skills.sort(key=len, reverse=True)

    final_skills = []

    for skill in found_skills:
        if not any(skill in existing for existing in final_skills):
            final_skills.append(skill)

    return final_skills
```

```python
required_skills = extract_required_skills(job_description)

print(required_skills)
```

```python
def extract_required_experience(text):
    patterns = [
        r'(\d+(?:\.\d+)?)\+?\s+years?\s+of\s+experience',
        r'(\d+(?:\.\d+)?)\+?\s+years?\s+experience',
        r'(\d+(?:\.\d+)?)\+?\s+years?\s+in'
    ]

    for pattern in patterns:
        match = re.search(pattern, text, flags=re.IGNORECASE)

        if match:
            return float(match.group(1))

    return 0.0
```

```python
required_experience = extract_required_experience(job_description)

print("Required Experience:", required_experience, "years")
```

```python
education_patterns = [
    r"\bBachelor'?s degree\b",
    r"\bMaster'?s degree\b",
    r"\bB\.Tech\b",
    r"\bB\.E\.\b",
    r"\bB\.Sc\.\b",
    r"\bB\.A\.\b",
    r"\bB\.Com\.\b",
    r"\bM\.Tech\b",
    r"\bM\.E\.\b",
    r"\bM\.Sc\.\b",
    r"\bMBA\b",
    r"\bMCA\b",
    r"\bPh\.D\.\b",
    r"\bBachelor of [A-Za-z ]+\b",
    r"\bMaster of [A-Za-z ]+\b"
]
```

```python
def extract_required_education(text):
    education = []

    for pattern in education_patterns:
        matches = re.findall(
            pattern,
            text,
            flags=re.IGNORECASE
        )

        for match in matches:
            match = match.strip()

            if match.lower() not in [
                item.lower() for item in education
            ]:
                education.append(match)

    return education
```

```python
required_education = extract_required_education(job_description)

print("Required Education:", required_education)
```

```python
job_analysis = {
    "job_description": job_description,
    "required_skills": required_skills,
    "required_experience": required_experience,
    "required_education": required_education
}

job_analysis
```

```python
job_analysis_df = pd.DataFrame([job_analysis])

job_analysis_df
```

```python
job_analysis_df.to_csv(
    "../data/processed/job_description_analysis.csv",
    index=False
)
```
